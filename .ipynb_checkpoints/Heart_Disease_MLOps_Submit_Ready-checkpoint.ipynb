## 1. Environment and project folders
The workflow uses fixed random seeds and writes every result to a predictable folder so that notebook outputs, report figures and deployment assets remain consistent.

```python
from pathlib import Path
import sys, json, warnings
import numpy as np, pandas as pd, matplotlib.pyplot as plt, joblib, sklearn
from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, classification_report, confusion_matrix, roc_curve, auc
warnings.filterwarnings('ignore')
for d in ['data','models','artifacts','screenshots','mlruns']: Path(d).mkdir(parents=True,exist_ok=True)
print('Python:',sys.version.split()[0]); print('NumPy:',np.__version__); print('Pandas:',pd.__version__); print('Scikit-learn:',sklearn.__version__); print('Working directory:',Path.cwd())
```

## 2. Data acquisition and validation
The processed Cleveland file has 13 predictors plus a diagnosis field. The diagnosis is converted from `0–4` into a binary target: `0` means absence and `1–4` mean presence. Missing markers are interpreted as null values and handled later inside the pipeline.

```python
columns=['age','sex','cp','trestbps','chol','fbs','restecg','thalach','exang','oldpeak','slope','ca','thal','num']
local=Path('data/processed.cleveland.data')
url='https://archive.ics.uci.edu/ml/machine-learning-databases/heart-disease/processed.cleveland.data'
try:
    df=pd.read_csv(local if local.exists() else url,header=None,names=columns,na_values='?')
except Exception as exc:
    raise RuntimeError('Download processed.cleveland.data from the UCI Heart Disease page and upload it to data/processed.cleveland.data, then rerun this cell.') from exc
df['target']=(pd.to_numeric(df.pop('num'),errors='coerce')>0).astype(int)
for col in df.columns: df[col]=pd.to_numeric(df[col],errors='coerce')
df.to_csv('data/heart_clean.csv',index=False)
print('Shape:',df.shape); print('Duplicates:',int(df.duplicated().sum())); print('Missing:',int(df.isna().sum().sum())); display(df.head(),df.describe().T)
```

## 3. Data quality and exploratory analysis
The analysis checks class balance, missingness, distributions and linear relationships. Figures are saved automatically for the final report.

```python
quality=pd.DataFrame({'dtype':df.dtypes.astype(str),'missing':df.isna().sum(),'missing_percent':(df.isna().mean()*100).round(2),'unique':df.nunique()})
quality.to_csv('artifacts/data_quality.csv'); display(quality)
counts=df.target.value_counts().sort_index(); print('Class counts'); print(counts); print(); print('Class percentages'); print((df.target.value_counts(normalize=True).sort_index()*100).round(2))
```

```python
fig,ax=plt.subplots(2,2,figsize=(13,9))
ax[0,0].bar(['No disease','Disease'],counts.values,color=['#2F7ED8','#D9534F']); ax[0,0].set_title('Target Class Distribution')
for t,c,l in [(0,'#2F7ED8','No disease'),(1,'#D9534F','Disease')]: ax[0,1].hist(df.loc[df.target==t,'age'].dropna(),bins=12,alpha=.55,color=c,label=l)
ax[0,1].set_title('Age Distribution by Target'); ax[0,1].legend()
ax[1,0].boxplot([df.loc[df.target==0,'thalach'].dropna(),df.loc[df.target==1,'thalach'].dropna()],labels=['No disease','Disease']); ax[1,0].set_title('Maximum Heart Rate')
ax[1,1].boxplot([df.loc[df.target==0,'oldpeak'].dropna(),df.loc[df.target==1,'oldpeak'].dropna()],labels=['No disease','Disease']); ax[1,1].set_title('Exercise ST Depression')
fig.suptitle('Heart Disease Dataset: Exploratory Overview'); fig.tight_layout(); fig.savefig('screenshots/eda_overview.png',dpi=180,bbox_inches='tight'); plt.show()
```

```python
corr=df.corr(numeric_only=True); fig,ax=plt.subplots(figsize=(12,9)); image=ax.imshow(corr,cmap='coolwarm',vmin=-1,vmax=1)
ax.set_xticks(range(len(corr.columns))); ax.set_yticks(range(len(corr.columns))); ax.set_xticklabels(corr.columns,rotation=90); ax.set_yticklabels(corr.columns); fig.colorbar(image,ax=ax,label='Correlation'); ax.set_title('Correlation Heatmap'); fig.tight_layout(); fig.savefig('screenshots/correlation_heatmap.png',dpi=180,bbox_inches='tight'); plt.show()
print('Strongest absolute target correlations'); print(corr.target.drop('target').abs().sort_values(ascending=False).head(5).round(3))
```

## 4. Leakage-safe preprocessing
Numerical variables use median imputation and standardisation. Categorical variables use most-frequent imputation and one-hot encoding. Preprocessing and estimation remain in one serialisable Scikit-learn pipeline.

```python
X=df.drop(columns='target'); y=df.target.astype(int)
cat=[c for c in ['sex','cp','fbs','restecg','exang','slope','ca','thal'] if c in X]; num=[c for c in X if c not in cat]
pre=ColumnTransformer([('numeric',Pipeline([('imputer',SimpleImputer(strategy='median')),('scaler',StandardScaler())]),num),('categorical',Pipeline([('imputer',SimpleImputer(strategy='most_frequent')),('encoder',OneHotEncoder(handle_unknown='ignore'))]),cat)])
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42,stratify=y); cv=StratifiedKFold(5,shuffle=True,random_state=42)
print('Train:',Xtr.shape,'Test:',Xte.shape); print('Numerical:',num); print('Categorical:',cat)
```

## 5. Model development, tuning and experiment tracking
Logistic Regression provides an interpretable baseline, while Random Forest captures non-linear interactions. Both use the same folds and held-out test set. Every run is written to CSV; MLflow is additionally used when installed.

```python
spaces={'Logistic Regression':(LogisticRegression(max_iter=2000,random_state=42),{'classifier__C':[.1,1,10]}),'Random Forest':(RandomForestClassifier(random_state=42,class_weight='balanced'),{'classifier__n_estimators':[100,200],'classifier__max_depth':[4,8,None]})}
try:
 import mlflow,mlflow.sklearn; MLFLOW=True; mlflow.set_tracking_uri('file:'+str((Path.cwd()/'mlruns').resolve())); mlflow.set_experiment('heart-disease-final'); print('MLflow enabled')
except Exception as exc: MLFLOW=False; print('MLflow unavailable; CSV ledger remains active:',exc)
```

```python
rows=[]; fitted={}
for name,(est,grid) in spaces.items():
 search=GridSearchCV(Pipeline([('preprocessor',pre),('classifier',est)]),grid,scoring='roc_auc',cv=cv,n_jobs=-1,refit=True).fit(Xtr,ytr)
 model=search.best_estimator_; pred=model.predict(Xte); prob=model.predict_proba(Xte)[:,1]
 metrics={'cv_roc_auc':float(search.best_score_),'test_accuracy':float(accuracy_score(yte,pred)),'test_precision':float(precision_score(yte,pred,zero_division=0)),'test_recall':float(recall_score(yte,pred,zero_division=0)),'test_f1':float(f1_score(yte,pred,zero_division=0)),'test_roc_auc':float(roc_auc_score(yte,prob))}
 if MLFLOW:
  with mlflow.start_run(run_name=name):
   mlflow.log_params({k.replace('classifier__',''):v for k,v in search.best_params_.items()}); mlflow.log_metrics(metrics); mlflow.sklearn.log_model(model,'model')
 rows.append({'model':name,**metrics,'best_parameters':json.dumps(search.best_params_)}); fitted[name]=(model,metrics,pred,prob)
comparison=pd.DataFrame(rows).sort_values('test_roc_auc',ascending=False).reset_index(drop=True); comparison.to_csv('artifacts/model_comparison.csv',index=False); display(comparison.round(4))
```

```python
metric_columns=['cv_roc_auc','test_accuracy','test_precision','test_recall','test_f1','test_roc_auc']; ax=comparison.set_index('model')[metric_columns].T.plot(kind='bar',figsize=(12,6),ylim=(0,1)); ax.set_title('Model Performance Comparison'); ax.set_ylabel('Score'); ax.tick_params(axis='x',rotation=35); plt.tight_layout(); plt.savefig('screenshots/model_comparison.png',dpi=180,bbox_inches='tight'); plt.show()
```

## 6. Final model evaluation and packaging
The final model is selected by held-out ROC-AUC while the complete metric profile remains visible. The saved object includes preprocessing and prediction, reducing training-serving skew.

```python
winner=comparison.loc[0,'model']; model,metrics,pred,prob=fitted[winner]; joblib.dump(model,'models/final_model.joblib'); Path('artifacts/final_metrics.json').write_text(json.dumps({'selected_model':winner,**metrics},indent=2))
cm=confusion_matrix(yte,pred); fpr,tpr,_=roc_curve(yte,prob); fig,ax=plt.subplots(1,2,figsize=(12,5)); ax[0].imshow(cm,cmap='Blues')
for i in range(2):
 for j in range(2): ax[0].text(j,i,cm[i,j],ha='center',va='center',fontsize=14)
ax[0].set_title('Confusion Matrix'); ax[0].set_xlabel('Predicted'); ax[0].set_ylabel('Actual'); ax[1].plot(fpr,tpr,label=f'ROC-AUC={auc(fpr,tpr):.3f}',color='#D9534F'); ax[1].plot([0,1],[0,1],'--',color='grey'); ax[1].set_title('ROC Curve'); ax[1].legend(); fig.suptitle('Final Model: '+winner); fig.tight_layout(); fig.savefig('screenshots/final_evaluation.png',dpi=180,bbox_inches='tight'); plt.show(); print('Selected:',winner); print(json.dumps(metrics,indent=2)); print(); print(classification_report(yte,pred))
```

## 8. Submission validation
Every item below should report `OK`. These artefacts are the evidence layer for the report, API, tests and deployment.

```python
required=['data/heart_clean.csv','models/final_model.joblib','artifacts/data_quality.csv','artifacts/model_comparison.csv','artifacts/final_metrics.json','screenshots/eda_overview.png','screenshots/correlation_heatmap.png','screenshots/model_comparison.png','screenshots/final_evaluation.png']; ok=True
for item in required:
 exists=Path(item).exists(); print('OK' if exists else 'MISSING',item); ok=ok and exists
assert ok and len(comparison)==2 and comparison[metric_columns].notna().all().all(); print(); print('Submission artefact validation passed.')
```

```python
from pathlib import Path

script = r'''
from pathlib import Path
import json
import sys

import pandas as pd
import joblib
import mlflow
import mlflow.sklearn


# ============================================================
# CONFIGURATION
# ============================================================

PROJECT_ROOT = Path.cwd()
EXPERIMENT_NAME = "heart-disease-final"

MLRUNS_DIR = (PROJECT_ROOT / "mlruns").resolve()

COMPARISON_FILE = (
    PROJECT_ROOT
    / "artifacts"
    / "model_comparison.csv"
)

FINAL_METRICS_FILE = (
    PROJECT_ROOT
    / "artifacts"
    / "final_metrics.json"
)

MODEL_FILE = (
    PROJECT_ROOT
    / "models"
    / "final_model.joblib"
)


# ============================================================
# VALIDATE FILES
# ============================================================

print("=" * 70)
print("HEART DISEASE MLOPS - MLFLOW SETUP")
print("=" * 70)

print("\nProject root:")
print(PROJECT_ROOT)


required_files = [
    COMPARISON_FILE,
    FINAL_METRICS_FILE,
    MODEL_FILE
]


missing = [
    path
    for path in required_files
    if not path.exists()
]


if missing:

    print("\nERROR - required files missing:")

    for path in missing:
        print("MISSING:", path)

    sys.exit(1)


for path in required_files:
    print("OK:", path)


# ============================================================
# CONFIGURE MLFLOW
# ============================================================

MLRUNS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


tracking_uri = MLRUNS_DIR.as_uri()


mlflow.set_tracking_uri(
    tracking_uri
)


experiment = mlflow.set_experiment(
    EXPERIMENT_NAME
)


print("\nMLflow Tracking URI:")
print(mlflow.get_tracking_uri())

print("\nExperiment:")
print(EXPERIMENT_NAME)

print("\nExperiment ID:")
print(experiment.experiment_id)


# ============================================================
# LOAD MODEL COMPARISON
# ============================================================

comparison = pd.read_csv(
    COMPARISON_FILE
)


print("\nMODEL COMPARISON")
print("-" * 70)

print(
    comparison.to_string(
        index=False
    )
)


# ============================================================
# FIND EXISTING RUNS
# ============================================================

existing = mlflow.search_runs(
    experiment_ids=[
        experiment.experiment_id
    ]
)


if len(existing) > 0:

    existing_names = set(
        existing[
            "tags.mlflow.runName"
        ].dropna()
    )

else:

    existing_names = set()


print("\nExisting runs:")

if existing_names:

    for name in sorted(existing_names):
        print("-", name)

else:
    print("None")


# ============================================================
# LOG MODEL EXPERIMENTS
# ============================================================

metric_columns = [
    "cv_roc_auc",
    "test_accuracy",
    "test_precision",
    "test_recall",
    "test_f1",
    "test_roc_auc"
]


for _, row in comparison.iterrows():

    model_name = str(
        row["model"]
    )


    if model_name in existing_names:

        print(
            "\nSKIPPING existing run:",
            model_name
        )

        continue


    print(
        "\nLogging:",
        model_name
    )


    with mlflow.start_run(
        run_name=model_name
    ):

        mlflow.log_param(
            "model_name",
            model_name
        )


        # Hyperparameters
        if "best_parameters" in comparison.columns:

            try:

                params = json.loads(
                    row[
                        "best_parameters"
                    ]
                )

                for key, value in params.items():

                    clean_key = (
                        key.replace(
                            "classifier__",
                            ""
                        )
                    )

                    mlflow.log_param(
                        clean_key,
                        value
                    )

            except Exception as exc:

                print(
                    "Parameter warning:",
                    exc
                )


        # Metrics
        for metric in metric_columns:

            if metric not in comparison.columns:
                continue

            value = row[metric]

            if pd.isna(value):
                continue

            mlflow.log_metric(
                metric,
                float(value)
            )


        # Comparison evidence
        mlflow.log_artifact(
            str(COMPARISON_FILE)
        )


print(
    "\nCandidate model logging completed."
)


# ============================================================
# FINAL MODEL
# ============================================================

with open(
    FINAL_METRICS_FILE,
    "r",
    encoding="utf-8"
) as file:

    final_metrics = json.load(
        file
    )


selected_model = final_metrics[
    "selected_model"
]


final_model = joblib.load(
    MODEL_FILE
)


print("\nSelected final model:")
print(selected_model)


# ============================================================
# LOG FINAL MODEL
# ============================================================

FINAL_RUN = "Final Model Evidence"


existing = mlflow.search_runs(
    experiment_ids=[
        experiment.experiment_id
    ]
)


existing_names = set(
    existing[
        "tags.mlflow.runName"
    ].dropna()
)


if FINAL_RUN not in existing_names:

    print(
        "\nLogging final model evidence..."
    )


    with mlflow.start_run(
        run_name=FINAL_RUN
    ):

        mlflow.log_param(
            "selected_model",
            selected_model
        )


        for key, value in final_metrics.items():

            if key == "selected_model":
                continue

            try:

                mlflow.log_metric(
                    key,
                    float(value)
                )

            except Exception:
                pass


        # Final serialised model
        mlflow.sklearn.log_model(
            final_model,
            "model"
        )


        # Assignment evidence
        artifacts = [

            "artifacts/model_comparison.csv",

            "artifacts/final_metrics.json",

            "artifacts/data_quality.csv",

            "screenshots/model_comparison.png",

            "screenshots/final_evaluation.png",

            "screenshots/correlation_heatmap.png",

            "screenshots/eda_overview.png"
        ]


        print("\nLogging artefacts:")


        for artifact_name in artifacts:

            artifact_path = (
                PROJECT_ROOT
                / artifact_name
            )


            if artifact_path.exists():

                mlflow.log_artifact(
                    str(artifact_path)
                )

                print(
                    "OK:",
                    artifact_name
                )

            else:

                print(
                    "NOT FOUND:",
                    artifact_name
                )


else:

    print(
        "\nFinal Model Evidence already exists."
    )


# ============================================================
# FINAL VERIFICATION
# ============================================================

experiment = (
    mlflow.get_experiment_by_name(
        EXPERIMENT_NAME
    )
)


runs = mlflow.search_runs(
    experiment_ids=[
        experiment.experiment_id
    ]
)


print("\n" + "=" * 70)
print("MLFLOW RUNS")
print("=" * 70)


columns = [

    "run_id",

    "tags.mlflow.runName",

    "metrics.cv_roc_auc",

    "metrics.test_accuracy",

    "metrics.test_precision",

    "metrics.test_recall",

    "metrics.test_f1",

    "metrics.test_roc_auc"
]


available_columns = [

    column
    for column in columns
    if column in runs.columns
]


print(
    runs[
        available_columns
    ].to_string(
        index=False
    )
)


run_names = set(
    runs[
        "tags.mlflow.runName"
    ].dropna()
)


expected = {

    "Logistic Regression",

    "Random Forest",

    "Final Model Evidence"
}


missing_runs = (
    expected
    - run_names
)


print("\n" + "=" * 70)


if missing_runs:

    print(
        "WARNING - missing runs:"
    )

    for run in missing_runs:
        print("-", run)

else:

    print(
        "PASS - ALL EXPECTED "
        "MLFLOW RUNS CREATED"
    )


print("=" * 70)


print("\nExperiment:")
print(EXPERIMENT_NAME)

print("\nMLflow directory:")
print(MLRUNS_DIR)

print(
    "\nRun this command next:"
)

print()

print(
    "mlflow ui "
    "--backend-store-uri ./mlruns "
    "--host 0.0.0.0 "
    "--port 5000"
)
'''

Path(
    "setup_mlflow_evidence.py"
).write_text(
    script,
    encoding="utf-8"
)

print(
    "Created:",
    Path(
        "setup_mlflow_evidence.py"
    ).resolve()
)

print(
    "Exists:",
    Path(
        "setup_mlflow_evidence.py"
    ).exists()
)
```
